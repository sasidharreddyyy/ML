```python
import math

data = [
    (40, 20, "Red"),
    (50, 50, "Blue"),
    (60, 90, "Blue"),
    (10, 25, "Red"),
    (70, 70, "Blue"),
    (60, 10, "Red"),
    (25, 80, "Blue")
]

new_brightness = 40
new_saturation = 40

for brightness, saturation, class_name in data:
    distance = math.sqrt(
        (brightness - new_brightness)**2 +
        (saturation - new_saturation)**2
    )

    print(class_name, round(distance, 2))
```
