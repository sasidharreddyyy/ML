```python
import numpy as np
```

```python
variables = [
    "Climate",
    "Housing",
    "Health",
    "Crime",
    "Transportation",
    "Education",
    "Arts",
    "Recreation",
    "Economy"
]
```

```python
loadings = np.array([
    [0.190,  0.017,  0.207],
    [0.544,  0.020,  0.204],
    [0.782, -0.605,  0.144],
    [0.365,  0.294,  0.585],
    [0.585,  0.085,  0.234],
    [0.394, -0.273,  0.027],
    [0.985,  0.126, -0.111],
    [0.520,  0.402,  0.519],
    [0.142,  0.150,  0.239]
])
```

```python
absolute_loadings = np.abs(loadings)

print(absolute_loadings)
```

```python
max_indices = np.argmax(absolute_loadings, axis=0)

print(max_indices)
```

```python
for i in range(3):
    index = max_indices[i]
    print(
        "PC", i + 1,
        "=", variables[index],
        "Loading =", loadings[index, i]
    )
```
